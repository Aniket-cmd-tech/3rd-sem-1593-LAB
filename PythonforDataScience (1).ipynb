# Definition

A tuple is ordered collection of items . It is immutable . Tuple is given in the ()

```python
# Create a tuple
student=("Chirag",21,'CSE',8.7)
print(student)
```

```python
# Display the items by index
subjects=("Python","DSA","DAA","OS","SQL","Statistics")
print(subjects[0])
print(subjects[5])
```

```python
subjects=("Python","DSA","DAA","OS","SQL","Statistics")
subjects[2]="Algebra"
```

```python
#tuple slicing and length, find the item is present in tuple
subjects=("Python","Statistics","SQL","Machine Learning","Deep Learning")
print(subjects[1:4])
print("Number of subjects: ",len(subjects))
print("Deep Learning" in subjects)
print("Relational Algebra" in subjects)
```

```python
# nested tuple
student=(101,'Rahul',(85,98,76))
print(student)
print(student[2])
print(student[2][1])
```

```python
#create a tuple
student=(101,"Rahul","CSE",8.9)
print(student)
#tuple using index
print("Student information in first location: ",student[0])
print(student)
#negative index
print("Student information in first location: ",student[1])
print(student[-1])
```

```python
#slicing
subjects=("DSA","DSV","Statistics","SQL","Machine Learning")
print(subjects[1:3])
print(subjects[:5])
print(subjects[4:])
```

```python
#Nested Tuple
student=(101,"Rahul",(85,78,66))
print(student)
print(student[2])
print(student[2][1])
```

# SET
## unordered collection of items
## mutable
## allows the different data types
## it doesn't support for index
## it gives output with unique(No duplication)

```python
# Symbol for set { }
# Create a set with similar data items
marks={75,68,89,55,78,89,75,66,55,99}
print(marks)
```

```python
#create a empty set
a={}
a=set()
#check its type
print(type(a))
```

```python
#set doesn't support the index
students={"Amit","Riya","Priya"}
print(students[1])
```

```python
# add an item in the set
students={"Amit","kiran","Sneha"}
students.add("Priya")
print(students)
#to add multiple values or iten
students.update(["Ram","Chitra","Vani","Devi"])
print(students)
```

```python
# remove an item from the set
students={"Amit","Riya","Priya"}
students.remove("Riya")
print(students)
students.discard("Amit")
print(students)
```

```python
students={ "Amit","Riya","Priya"}
students.remove("Karan")
print(students)
```

```python
students={"Amit","Priya","Riya"}
students.discard("Karan")
print(students)
```

```python
# create a set and clear a set of list of item
students_info={"Amit","karan","Riya"}
print(students_info)
students_info.clear()
print(students_info)
```

```python
marks={99,88,77,66}
del marks
print(marks)
```

```python
# set operations
python_students={"Amit","Neha","Sanjiv","Manish"}
java_students={"Amit","karan","Dilip","Manish","Neha"}
# for union operations by using set
result=python_students.union(java_students)
print("union operation: ",result)
# for intersection operations by using set
result=python_students.intersection(java_students)
print("intersection operation:",result)
# for difference operations by using set
result=python_students.difference(java_students)
print("difference operation:",result)
```

```python
# convert list to set
stud_id=[101,102,103,102,103,104,105]
Unique_stud_id=set(stud_id)
print(Unique_stud_id)

unique_roll_no=list(Unique_stud_id)
print(unique_roll_no)
```
