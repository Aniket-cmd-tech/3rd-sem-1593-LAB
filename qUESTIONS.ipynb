File Formats
1. csv
2. excel
3. json
4. parquet


```python
# Load the dataset in the file format
Iris.to_csv("Iris_csvoutput.csv",index=False)
#Read the saved dataset
Iris_csv=pd.read_csv("/content/Iris_csvoutput.csv")
print(Iris_csv.head())
```

```python
Iris.to_excel("Iris_exceloutput.xlsx",index=False)
#Load
Iris_excel=pd.read_excel("/content/Iris_exceloutput.xlsx")
print(Iris_excel.tail(10))
```

```python
#json file format
Iris.to_json("iris_jsonoutput.json",orient="records",indent=4)
#Read the file
Iris_json=pd.read_json("/content/iris_jsonoutput.json")
Iris_json.head(5)
```

```python
#parquet file format
Iris.to_parquet("Iris_parquetoutput.parquet",index=False)
#read the file
Iris_parquet=pd.read_parquet("/content/Iris_parquetoutput.parquet")
print(Iris_parquet.head(3))
```

```python

# Comparing the size of the file formats
import os

files = [
    "Iris_csvoutput.csv",
    "Iris_exceloutput.xlsx",
    "Iris_jsonoutput.json",
    "Iris_parquetoutput.parquet"
]

for file in files:
    size = os.path.getsize(file)
    print(f"{file}: {size} bytes")
```
