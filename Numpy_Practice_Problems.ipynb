# 1. Array Creation & Properties

1. Create a NumPy array from [10, 20, 30, 40, 50]. Display its shape, size, and data type.

```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
print(arr.shape)
print(arr.size)
print(arr.dtype)
```

```python
import numpy as np
arr = np.arange(1, 13).reshape(3, 4)
print(arr)
```

3. Create an array of zeros, an array of ones, and an identity matrix.

```python
import numpy as np
zeros_arr = np.zeros((3, 3))
ones_arr = np.ones((3, 3))
identity_arr = np.eye(3)
print(zeros_arr)
print(ones_arr)
print(identity_arr)
```

# 2. Array Arithmetic

6. Given two NumPy arrays, perform element-wise addition, subtraction, multiplication,
and division.

```python
import numpy as np
arr1 = np.array([10, 20, 30, 40])
arr2 = np.array([2, 4, 5, 8])
print(arr1 + arr2)
print(arr1 - arr2)
print(arr1 * arr2)
print(arr1 / arr2)
```

7. Calculate the square and cube of every element in a NumPy array.

```python
import numpy as np
arr = np.array([1, 2, 3, 4, 5])
print(np.square(arr))
print(np.power(arr, 3))
```

8. Increase every element of an array by 15%.

```python
import numpy as np
arr = np.array([100.0, 200.0, 300.0])
updated_arr = arr * 1.15
print(updated_arr)
```

9. Given two arrays, find their element-wise maximum and minimum.

```python
import numpy as np
import numpy as np
arr1 = np.array([3, 8, 2, 10])
arr2 = np.array([5, 4, 9, 1])
print(np.maximum(arr1, arr2))
print(np.minimum(arr1, arr2))
```

# 3. Indexing & Slicing

10. Create an array from 0 to 20 and extract the first five elements, last five elements,
elements from index 5 to 15, and every second element.

```python
import numpy as np
arr = np.arange(21)
print(arr[:5])
print(arr[-5:])
print(arr[5:16])
print(arr[::2])
```

11. Replace the first five elements of an array with 100.

```python
import numpy as np
arr = np.arange(10)
arr[:5] = 100
print(arr)
```

12. Create an array and demonstrate the difference between a slice/view and a copy.

```python
import numpy as np
original = np.array([1, 2, 3, 4, 5])
view_arr = original[1:4]
view_arr[0] = 999
copy_arr = original[1:4].copy()
copy_arr[1] = 888
print(original)
print(copy_arr)
```

13. Reverse a NumPy array using slicing.

```python
import numpy as np
arr = np.array([1, 2, 3, 4, 5])
reversed_arr = arr[::-1]
print(reversed_arr)
```

# 4. 2-D Array Indexing & Slicing

14. Create the matrix [[10,20,30],[40,50,60],[70,80,90]]. Extract the second row, third
column, and element 50.

```python
import numpy as np

matrix = np.array([[10, 20, 30],
                   [40, 50, 60],
                   [70, 80, 90]])

print(matrix[1, :])
print(matrix[:, 2])
print(matrix[1, 1])
```

15. From the matrix above, extract the sub-matrix [[20,30],[50,60]].

```python
import numpy as np
matrix = np.array([[10, 20, 30],
                   [40, 50, 60],
                   [70, 80, 90]])
sub_matrix = matrix[0:2, 1:3]
print(sub_matrix)
```

16. Create a 5 × 5 matrix containing 1–25 and extract the first row, last column, middle
element, and top-left 3 × 3 matrix.

```python
import numpy as np
matrix_5x5 = np.arange(1, 26).reshape(5, 5)
print(matrix_5x5[0, :])
print(matrix_5x5[:, -1])
print(matrix_5x5[2, 2])
print(matrix_5x5[0:3, 0:3])
```

# 5. Reshaping & Transpose

17. Create an array containing 1–12 and reshape it into a 3 × 4 matrix.

```python
import numpy as np
arr = np.arange(1, 13).reshape(3, 4)
print(arr)
```

18. Reshape a 1-D array into a 2 × 5 matrix.

```python
import numpy as np
arr_1d = np.arange(10)
matrix_2x5 = arr_1d.reshape(2, 5)
print(matrix_2x5)
```

19. Convert a 3 × 4 matrix into a 4 × 3 matrix using transpose.

```python
import numpy as np
matrix_3x4 = np.arange(12).reshape(3, 4)
matrix_4x3 = matrix_3x4.T
print(matrix_4x3)
```

20. Use transpose() and swapaxes() on a 3-D array.

```python
import numpy as np
arr_3d = np.arange(24).reshape(2, 3, 4)
transposed = arr_3d.transpose(1, 0, 2)
swapped = arr_3d.swapaxes(0, 1)
print(transposed.shape)
print(swapped.shape)
```
