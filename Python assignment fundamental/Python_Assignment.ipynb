Answer in 2-3 sentences.

Q7.1

```python
def my_abs(val):
    if val < 0:
        return 0 - val
    return val
print(my_abs(7))
print(my_abs(-7))
print(my_abs(0))
```

Q7.5

```python
def swap_message(val1, val2):
    temp = val1
    val1 = val2
    val2 = temp
    print("Inside function:")
    print("val1 =", val1)
    print("val2 =", val2)
x = 6
y = 3
swap_message(x, y)
print("Outside function:")
print("x =", x)
print("y =", y)
```

Topic 8: Scope — Local vs Global Variables

Practice Questions:

Q8.1

```python
def my_abs(val):
    if val < 0:
        return 0 - val
    return val
print(val)
```

val is a local variable because it is created as a parameter of my_abs(). It only exists inside the function. When we try print(val) outside the function, Python cannot find val.

Q8.2

```python
counter = 0
def show_counter():
    print(counter)
show_counter()
```


counter is a global variable, so the function can read its value even though it was created outside the function.

Q8.3

```python
counter = 0
def increment_counter():
    counter = counter + 1
increment_counter()
```

Q8.4

```python
counter = 0
def increment_counter():
    global counter
    counter = counter + 1
increment_counter()
increment_counter()
increment_counter()
print(counter)
```
